# **Definimos la funcion dada**

```python
def funcion(x):
  return x**3 + 4*x**2 - 10
```

# Definamos los parametros iniciales

```python
p0 = 1.0
h0 = 2.0
tol = 1e-4
nmax = 100
```

# Hacemos la INICIALIZACION

```python
niter = 1
p = p0
h = h0
fp = funcion(p)
fh = funcion(h)
error = (h - p) / 2
```

# Se crea una lista para guardar los datos de la tabla

```python
datos = []
```

# Ciclo Iterativo

```python
import numpy as np
import pandas as pd

while error > tol and niter <= nmax:
  # Primero se calcula punto medio
  m = p + (h - p) / 2
  fm = funcion(m)

  #Guardamos los datos para la tabla
  datos.append({
       'n': niter,
       'p_n': p,
       'h_n': h,
       'q_n (m)': m,
       'f(q_n)': fm,
       'Error': error
  })

  # Segundo evaluamos el criterio de parada
  if fm == 0 or error < tol:
     break

  #Tercero actualizamos el intervalo
  if np.sign(fp) == np.sign(fm):
    p = m
    fp = fm
  else:
    h = m
    fh = fm

  #Cuarto actualizar el error
  error = (h - p) / 2
  niter += 1
```

# Hacemos la tabla

```python
df_tabla = pd.DataFrame(datos)
print("Tabla de iteraciones")
display(df_tabla)

print(f"\n La raiz aproximada es: {m:.9f}")
print(f"El valor de f(m) es: {fm:.9f}")
print(f"Iteraciones realizadas: {niter - 1}")
```

# Ahora creamos la grafica

```python
import matplotlib.pyplot as plt

x_vals = np.linspace(0.5, 2.5, 400)
y_vals = funcion(x_vals)

plt.figure(figsize=(10, 5))
plt.plot(x_vals, y_vals, label='$f(x) = x^3 + 4x^2 - 10$', color='blue')
plt.axhline(0, color='black', linestyle='--')
plt.axvline(m, color='red', linestyle=':', label=f'Raiz aprox = {m:.5f}')
plt.title('Metodo de Biseccion - Ejemplo')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.grid(True, linestyle='--', alpha=0.7)
plt.legend()
plt.show()
```
